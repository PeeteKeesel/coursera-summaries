## <a id='toc1_3_'></a>[Limited Liability](#toc0_)

- Divide up an enterprise into shares
  - and no shareholder is liable for more than he/she put in
- Since NY State 1811: <span style="color:brown">**$\textbf{limited liability}$**</span>
- $\Leftrightarrow$ <span style="color:blue">Investors in stocks can never be pursued for the mistakes of the managers of the company</span>
- Before: if you invested in the company, **you** were responsible
- Human experiment result: If you are responsible for everything you get involved in, then you won't do it, you won't supply capital to some business
- **Investors psychology favors limited liability**
  - Cause if there is no limited liability, investors tend to overestimate the minuscule probability of loss. 
- Only with limited liability, diversification of the portfolio makes sense

## <a id='toc2_1_'></a>[Forecasting](#toc0_)

- plausible path vs expected paths of e.g. the stock market
  - plausible: ups and downs, as seen in the past
  - expected: only up, since in the past the stock market on average went only up
- <span style="color:brown">**$\textbf{Random Walk Hypothesis}$**</span> (Pearson, 1905)
  - <span style="color:blue">A random walk is a process that changes in such a way that each change is independent of previous changes and totally unforecastable.</span>
- Some professors argue: "*Stock market prices are really random walks*"
  - So it's an illusion that you think you can forecast it

### <a id='toc2_1_1_'></a>[Random Walk & AR-1 Models](#toc0_)

**Random Walk**

$$
    x_t = x_{t-1} + \epsilon_t
$$

<span style="color:brown">**$\textbf{1st-order autoregressive (AR-1) Model}$**</span> ($=$ alternative to random walk model)

$$
    x_t = 100 + \rho ( x_{t-1} - 100 ) + \epsilon_t
$$

- this model is mean reverting (to $100$)
- $\rho \in (-1, 1)$

Random walk as **approximate implication of unpredictability of returns**

> If stock prices are an AR-1, then it means you should <span style="color:orange">stay out of the market when the price is too high and go back in when the price is too low</span>.

### <a id='toc2_1_3_'></a>[Visualization: Random Walk versus AR1](#toc0_)

```python
import matplotlib.pyplot as plt
import numpy as np

def generate_ar1_model(rho, c, num_steps):
    np.random.seed(123)
    random_walk = np.random.randn(num_steps)
    ar1_model = np.zeros(num_steps)
    ar1_model[0] = c
    for t in range(1, num_steps):
        ar1_model[t] = c + rho * (ar1_model[t-1] - c) + random_walk[t]
        
    return ar1_model

# --- 1. Generate Random Walk Data ---
np.random.seed(123)
num_steps = 200
random_walk = np.random.randn(num_steps)

# --- 2. Generate AR-1 Model Data ---
ar1_model = generate_ar1_model(rho=0.7, 
                               c=0, 
                               num_steps=num_steps)

ar1_model_0_9 = generate_ar1_model(rho=0.9, 
                                   c=0, 
                                   num_steps=num_steps)

ar1_model_5 = generate_ar1_model(rho=0.7, 
                                   c=5, 
                                   num_steps=num_steps)

# --- 3. Plotting ---
plt.figure(figsize=(12, 6))

plt.plot(random_walk, label='Random Walk', color='C0', linewidth=1, linestyle='-')
plt.plot(ar1_model, label=r'AR-1 Model ($\rho = 0.7$, c=0)', color='C1', linewidth=1, linestyle='-')
plt.plot(ar1_model_0_9, label=r'AR-1 Model ($\rho = 0.9$, c=0)', color='C2', linewidth=1, linestyle='-')
plt.plot(ar1_model_5, label=r'AR-1 Model ($\rho = 0.7$, c=5)', color='C3', linewidth=1, linestyle='-')
plt.legend()
plt.show()
```

## <a id='toc2_2_'></a>[Intuition of Efficiency](#toc0_)

  - **Semi-strong form**: prices incorporate *all publicly available* information
  - **Strong form**: prices incorporate **all** information, including insider information
- **buying stocks with low price-earning (PE) - ratios has paid off for a long time**

## <a id='toc2_3_'></a>[Price as PDV](#toc0_)

- If earnings $=$ dividends and if dividends grow at long-run rate $g$, then by <span style="color:brown">**$\textbf{growing consol model}$**</span> (Gordon Model): 

$$
\begin{align}
    P            &= \frac{E}{r-g} \\
    \mathbf{P/E} &= \frac{1}{r-g}
\end{align}
$$
where
- $P =$ Price
- $E =$ Earnings
- $r =$ Discount rate
- $g =$ growth rate (of earnings)

We can say:
- So, efficient markets theory purports to explain why PE varies across markets
- Low P/E **does not mean** that the stocks a "bargain"!
  - it means
    - either that earnings are rationally forecasted to decrease in future
    - or that it's a low risk stock
- Efficient markets denies that any rule works

Note, the price of a stock should be the present discounted value (PDV) of expected dividends.

If a stock has a very high P/E ration, e.g. $100$, it (should) mean that either:
1. the stock is very good in terms of risk ($r$), or
2. the growth rate of the earnings is going to be phenomenal ($g$)

So that **high price-earnings ratio *should* predict high growth rates of earnings**.
