## <a id='toc1_2_'></a>[Salon - Innovation](#toc0_)

- technology that we have has been very slow to develop over the centuries.
- financial innovation is a pillar of our civilization.
- people seem to like gambles that involve a small amount of money lost with high $\mathbb{P}$. And a large money gained with low $\mathbb{P}$.
  - So if that's what people like, we have to try to turn the natural risks which might be balanced equally into loss and gain into something more appealing.
- Part of the financial innovation:
  - Reframe risk so that they're appealing
  - and for good social purpose we want to have risks that *need* to be taken
  - so that they benefit the economy

## <a id='toc1_3_'></a>[Limited Liability](#toc0_)

- Divide up an enterprise into shares
  - and no shareholder is liable for more than he/she put in
- Since NY State 1811: <span style="color:brown">limited liability</span>
- $\Leftrightarrow$ <span style="color:blue">Investors in stocks can never be pursued for the mistakes of the managers of the company</span>
- Before: if you invested in the company, **you** were responsible
- Human experiment result: If you are responsible for everything you get involved in, then you won't do it, you won't supply capital to some business
- **Investors psychology favors limited liability**
  - Cause if there is no limited liability, investors tend to overestimate the minuscule probability of loss. 
- Only with limited liability, diversification of the portfolio makes sense

## <a id='toc1_4_'></a>[Inflation Indexed Debt](#toc0_)

- History shows many examples of nominal debt being wiped out in real terms by high inflation
- Idea: Why don't we have a price index and have a debt contract that pays back index to inflation? 
- You wanna tie the contract to an index
- Platon: <span style="color:blue">"Necessity is the mother of invention."</span>
- Issue inflation indexed debt
