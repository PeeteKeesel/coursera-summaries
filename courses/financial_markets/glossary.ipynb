$\textbf{Market Liquidity}$:
- "Can an individual or firm quickly purchase/sell an asset w/o drastic change in the assets price?"
  - Yes, $\Rightarrow$ liquid
    - Can be selled rapidly,
    - with minimal loss of value, 
    - anytime within market hours
  - No, $\Rightarrow$ non-liquid
- Liquidity involves the trade-off-between the <span style="color:blue">*price*</span> at which an asset can be sold, and <span style="color:blue">*how quickly*</span> it can be sold [[source](https://en.wikipedia.org/wiki/Market_liquidity)]
  - liquid market: trade-off $=$ low $\Leftrightarrow$ one can sell quickly w/o having to accept a significantly lower price
    - There are always ready and willing buyers and sellers
  - illiquid market: asset must be discounted in order to sell quickly
- Liquid asset $=$ convertable into cash in relatively short period of time
  - Cash $=$ most liquid asset

$\textbf{Bond}$ (Anleihe):
- type of security 
- under which the issuer (<span style="color:brown">debtor</span>) owes the holder (<span style="color:brown">creditor</span>) a debt
- Anleihen (im Englischen "Bonds" genannt) sind im Wesentlichen Schuldverschreibungen. Wenn Sie eine Anleihe kaufen, leihen Sie Geld an einen Emittenten – das kann ein Staat (z.B. Deutschland, USA), ein Unternehmen oder eine andere Organisation sein. Im Gegenzug verspricht Ihnen der Emittent, Ihnen Ihr geliehenes Geld zu einem bestimmten Zeitpunkt in der Zukunft (dem Fälligkeitsdatum) zurückzuzahlen.
- Während der Laufzeit der Anleihe erhalten Sie in der Regel regelmäßige Zinszahlungen, die man Kuponzahlungen nennt. Diese Kuponzahlungen sind der "Preis", den der Emittent dafür zahlt, dass er Ihr Geld nutzen darf.
- Man kann Anleihen also als eine Art Kredit vorstellen:
  - Sie sind der Gläubiger (der Geldgeber).
  - Der Emittent ist der Schuldner (der Geldnehmer).
Wichtige Merkmale von Anleihen:
1. **Nennwert** (Par Value / Face Value): Dies ist der Betrag, den der Emittent Ihnen am Fälligkeitsdatum zurückzahlt. Oft ist der Nennwert 1.000 Euro oder ein Vielfaches davon.
2. **Kuponzinssatz** (Coupon Rate): Der feste Zinssatz, der auf den Nennwert angewendet wird, um die Höhe der Kuponzahlungen zu bestimmen. Wenn eine Anleihe zum Beispiel einen Nennwert von 1.000 Euro und einen Kuponzinssatz von 3% hat, erhalten Sie jährlich 30 Euro Zinsen.
3. **Fälligkeitsdatum** (Maturity Date): Das Datum, an dem der Emittent den Nennwert der Anleihe zurückzahlt. Anleihen können sehr unterschiedliche Laufzeiten haben, von wenigen Monaten bis zu 30 Jahren oder länger.
4. **Emittent** (Issuer): Die Partei, die die Anleihe ausgibt und sich Geld leiht. Die Bonität des Emittenten (d.h. wie wahrscheinlich es ist, dass er seine Schulden zurückzahlen kann) ist entscheidend für das Risiko der Anleihe. Staaten mit hoher Bonität (wie Deutschland) gelten als sehr sicher, während bestimmte Unternehmen ein höheres Ausfallrisiko haben können.
5. **Rendite** (Yield): Die tatsächliche jährliche Ertragsrate einer Anleihe, die sowohl die Kuponzahlungen als auch etwaige Gewinne oder Verluste aus dem Kaufpreis (wenn die Anleihe über oder unter dem Nennwert gekauft wurde) berücksichtigt.

$\textbf{"Issued at par"}$: (or "at par")
- $=$ a bond or other security is sold for its face value (also known as par value or nominal value)
- When a bond is "issued at par," it means that the initial price at which it is sold to investors is exactly equal to its face value.
- This usually happens when the bond's coupon rate (the interest rate it pays) is equal to the prevailing market interest rate for similar bonds at the time of issuance. If the coupon rate is higher than market rates, the bond would typically be issued at a premium (above par). If the coupon rate is lower, it would be issued at a discount (below par).

$\textbf{Consumer Price Index}$ (CPI):
- $=$ measure of the average change over time in the prices paid by urban consumers for a "market basket" of consumer goods and services
  - "market basket" $=$ representative collection of goods and services that typical households buy, including things like food, housing, transportation, medical care, and apparel
    - The items and their weights in the basket are updated periodically to reflect changing consumer spending habits.
- key indicator of **inflation**, show how the cost of living changes
- **Calculation**: The CPI is calculated by comparing the current cost of this market basket to its cost in a base period. The percentage change in the CPI over time indicates the rate of inflation or deflation.
