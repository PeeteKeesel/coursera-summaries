- **EONIA (Europe)**: This is Europe's equivalent overnight interbank lending rate. Like the US, it has trended downwards since 2000. Interestingly, it has gone negative for over a year, even though holding physical cash yields 0%.

**Why *negative* interest rates?**

Banks lend to each other at negative rates because storing large amounts of physical cash is costly. This includes expenses for:
- Security (insurance, guards)
- Transportation (armored trucks)
- Lack of suitable infrastructure (banks aren't set up to hold billions in physical cash).

**Traditional Causes of Interest Rates**

According to Eugen Bohm von Bawerk (1884), interest rates are typically small positive numbers (like 3-5%) due to:
- **Technical Progress**: The economy grows over time, making future money more valuable (e.g., $1 today is like $1.03 in a year).
- **Time Preferences**: People are inherently impatient and prefer money now rather than later.
- **Advantages to "Roundaboutness"**: More delayed and complex production processes can yield greater returns (e.g., investing in fertilizer for more apples next year).

He said, interest rates are often ~3% or 5% because the *rate of progress* is about that.

**Low Interest Rates and Inequality**

Low interest rates can contribute to inequality. For example, elderly people relying on fixed incomes that are tied to interest earnings suffer when rates are near zero, as their income stream dries up. Monetary policy is a blunt tool and doesn't affect everyone equally.

## <a id='toc1_3_'></a>[Compound Interest](#toc0_)

- `Zinseszins`
- If annual rate is $r$, compounding *once* per year, balance $= (1 + r)^t$, after $t$ years
- If compounded *twice* per year, $= (1 + \frac{r}{2})^{2t}$, after $t$ years
- If compounded $n$ times per year, $= (1 + \frac{r}{n})^{nt}$, after $t$ years
- Continuous compounding (if $n \rightarrow \infty$), balance is $\exp^{rt}$

```python
import numpy as np
import matplotlib.pyplot as plt

annual_rate = 0.02 # 2%
t_years = np.arange(1, 11)
initial_balance = 100_000

balance_1x_per_year = initial_balance * (1 + annual_rate) ** t_years
balance_2x_per_year = initial_balance * (1 + annual_rate / 2) ** (2 * t_years)
balance_12x_per_year = initial_balance * (1 + annual_rate / 12) ** (12 * t_years)

print(f"Balance after 10 years with 2% annual rate:")
print(f"  - Once per year    : {balance_1x_per_year[-1]:,.2f}")
print(f"  - Twice per year   : {balance_2x_per_year[-1]:,.2f}")
print(f"  - 12 times per year: {balance_12x_per_year[-1]:,.2f}")

plt.plot(t_years, balance_1x_per_year, 'o-', color='blue', label='1x per year')
plt.plot(t_years, balance_2x_per_year, 'x-', color='red', label='2x per year')
plt.plot(t_years, balance_12x_per_year, '>-', color='green', label='12x per year')

plt.legend()
plt.xlabel('Time (years)')
plt.ylabel('Balance')
plt.xticks(t_years)
plt.title(f'Compound Interest with {100*annual_rate:.0f}% annual rate')
plt.show()
```

## <a id='toc1_4_'></a>[Discount Bounds](#toc0_)

- Bonds typically pay coupons. 
  - This is an old word but they still say coupons. 
  - It used to be, that if you invested in a corporate bond or a government bond it would be on a piece of paper they would give you. 
  - This is like for hundreds of years. 
  - And around the exterior of the pieces of paper, were little coupons that you would clip every six months typically. 
  - And you would clip your coupons every six months and take them to a bank, and the bank would then give you the money. 
  - So, each coupon would be so much money. 
  - And then, at the end of the maturity of the bond, you could take the whole thing back and you get your amount back. 

<span style="color:brown">**$\textbf{Discount Bounds}$**</span> $=$ Bond that carries no coupon.

- No coupon payments, just principal at maturity date (conventionally, $100)

Why would anyone take a bond with no coupon and thus no interest? 
- because you buy it at a discount

Initially sold at a discount (less than $100) and price rises through time, creating income
- **Term** $t$
- **Yield to Maturity** (YTM) $r$

$$ 
    P = \frac{1}{(1 + r)^T}
$$

- $P = $ price
- typically bonds pay interest rates every 6 months, thus the formula becomes 

$$
    P = \frac{1}{(1 + \frac{r}{2})^{2T}}
$$

which has bond compounding 2x per year.

There tend to be two different kinds of bonds, 
1. the **coupon bonds** which are more common, tend to be sold at par. You buy it initially for $100, and you sell it for a 100, you get back at the end when it matures after so many predefined years. 
2. with a **discount bond**, there are no coupons but, you buy it at a discount. There would be no other reason to buy it.

```python
# --- Bond Parameters ---
face_value = 1_000  # The amount the bondholder receives at maturity
time_to_maturity = 5  # Years until the bond matures (T)

# --- Yield to Maturity (YTM) Range ---
yield_to_maturity_range = np.linspace(0.01, 0.15, 100) # 1% to 15% with 100 points

# --- Calculate Discount Bond Price for each YTM ---
# For a discount (zero-coupon) bond, the price (P) is calculated as:
# P = Face Value / (1 + YTM)^T
# Where:
#   Face Value = the bond's par value (paid at maturity)
#   YTM = Yield to Maturity (discount rate)
#   T = Time to Maturity (in years)
bond_prices = face_value / (1 + yield_to_maturity_range)**time_to_maturity

# --- Plotting the Relationship ---
plt.figure(figsize=(10, 6))
plt.plot(yield_to_maturity_range, bond_prices, marker='o', markersize=4, linestyle='-', color='skyblue')

plt.title(f'Discount Bond Price vs. Yield to Maturity (FV=${face_value}, T={time_to_maturity} years)', fontsize=14)
plt.xlabel('Yield to Maturity (YTM)', fontsize=12)
plt.ylabel('Bond Price ($)', fontsize=12)

# Format x-axis as percentages
plt.gca().xaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: '{:.0%}'.format(y)))

# Add a grid for easier reading
plt.grid(True, linestyle='--', alpha=0.7)

plt.text(0.02, 800, 'Higher YTM = Lower Price', fontsize=10, color='red')
plt.text(0.12, 500, 'Lower YTM = Higher Price', fontsize=10, color='green', ha='right')

plt.show()

print(f"Plot generated for a discount bond with Face Value = ${face_value} and Time to Maturity = {time_to_maturity} years.")
print("The x-axis represents the Yield to Maturity, and the y-axis represents the Bond Price.")
print("Notice the inverse relationship: as YTM increases, the bond price decreases.")
```

<span style="color:brown">**$\textbf{Present Discounted Value}$**</span> (PDV)
- PDV of a dollar in $1$ year $= \frac{1}{(1 + r)}$
  - $r =$ yield to maturity on a 1-year discount bond
- PDV of a dollar in $n$ year $= \frac{1}{(1 + r)^n}$ 
- PDV of a stream of payments $x_1, \dots, x_n$
  - $x_1 =$ money coming in in $1$ year
  - $x_2 =$ money coming in in $2$ years
  - etc.

$$
    PDV = \sum_{t}^{T} \frac{x_t}{(1 + r)^t}
$$
- $x_t =$ cash flow received at time $t$
- $t =$ specific time period (e.g. year $1$, year $2$, etc.)
- $T =$ total number of periods until the last cash flow.
- $r =$ constant discount rate.

In general
- the PDV is of any cash flow, not just a coupon flow which would be the case for a coupon bond

$\textbf{Conventional Bonds Carry Coupons}$
- Conventional Bond issued at par (100), coupons every six months
- Term is time to maturity

$$
\begin{align}
    P_t &= c(\frac{1}{r} - \frac{1}{(1 + r)^T}\frac{1}{r}) + \frac{100}{(1 + r)^T} \qquad &&{\color{grey}{\text{compounded anually}}}\\
    P_t &= \frac{c}{2}( \frac{1}{r/2} - \frac{1}{(1 + r/2)^{2T}}\frac{1}{r/2}) + \frac{100}{(1 + r/2)^{2T}} \qquad &&{\color{grey}{\text{compounded every six months}}}
\end{align}
$$

- $c =$ amount of coupon

Every year (or six months in case of equation (2)) I receive my coupon value. And in the very end after maturity I receive additionally the initial value, the principal. Whats the present value of that, if discounted by $r$? The above equations.

## <a id='toc1_5_'></a>[Consol and Annuity](#toc0_)

- <span style="color:brown">**$\textbf{Consol}$**</span> pays constant quantity $x$ forever
- <span style="color:brown">**$\textbf{Growing Consol}$**</span> pays $x(1 + g)^{(t-1)}$ in $t$

$$
\begin{align}
    \text{Consol PDV}         &= \frac{x}{r} \\
    \text{Growing Consol PDV} &= \frac{x}{(r - g)}
\end{align}
$$

- $g =$ percent of growth per year

Why do we call them consols? 
- Because in Britain, in the 1700s, the government of the United Kingdom issued bonds with no maturity date. They promised to pay this coupon forever, and the only way they could get out of it was by buying them back. And those bonds, well, there was some adjustments made in the 1880s. They're still paying the coupon. The British government hasn't defaulted in all that time. And so it's not forever, but it's close to forever. I mean, several hundred years, a long time to be paying a coupon.


<span style="color:brown">**$\textbf{Annuity}$**</span> pays $x$ from time $1$ to $T$

$$
\begin{align}
    \text{Annuity PDV} = x \frac{1 - \frac{1}{(1+r)^T}}{r}
\end{align}
$$

Remember
- in bonds, the coupon is **fixed** at the time the bond is issued but the market price of the bond changes through time

> **Bonds are risky, have market risk, even if there is no default risk.**

- If the interest rate is $0$, then consol's don't even converge

Summary:

- **Consols**: These are bonds that pay a fixed coupon payment indefinitely, with no maturity date. The British government issued them in the 1700s. Their present value (price) is calculated as `Coupon / Interest Rate (C/r)`.
  - **Market Risk**: Even without default risk, consols (and other long-term debt instruments) have market risk because their price fluctuates with market interest rates. The coupon is fixed, but if market rates change, the bond's price adjusts, changing its yield to maturity.
- **Growing Consols**: A variation where the coupon payment grows at a constant rate ($g$) forever. The present value is `Coupon / (r - g)`.
  - **Important Note**: This formula breaks down (becomes infinite) if the growth rate ($g$) is equal to or greater than the discount rate ($r$), which highlights why interest rates must generally be positive for assets with growing payments to have a finite price.
- **Annuities**: Similar to consols, but payments *stop* after a certain number of years. 
  - A common example is a home mortgage, where regular payments are made over a set period, and there's no large principal repayment at the end (unlike a typical corporate bond). These instruments are designed to accommodate human behavior, making repayments easier (e.g., monthly payments instead of large balloon payments).

The growing consol formula goes back to Bernoulli.

## <a id='toc1_6_'></a>[Forward Rates and Expectation Theory](#toc0_)

<span style="color:brown">**$\textbf{Forward Rates}$**</span> $=$ interest rates that can be taken in advance using term structure
- J.R. Hicks, *Value and Capital*, 1939:
$$
\begin{align*}
    (1 + r_2)^2 &= (1 + r_1)(1 + f_2) \\ 
    (1 + r_k)^k &= (1 + r_{k-1})^{k-1}(1 + f_k) \\ 
\end{align*}
$$

<p align="center">
  <img width="600" height="300" src="imgs/03_module/m3l8-forward-rates.png">
</p>

- **Forward Rates:** These are interest rates for future periods that can be "locked in" today using the current "term structure" of interest rates.
- **Origin Story:** The speaker tried to discover who invented the concept of forward rates, even writing to Sir John Hicks (a prominent economist). Hicks couldn't definitively recall, suggesting it might have emerged from informal conversations at the London School of Economics in the 1920s.
- **How to "Lock In" a Forward Rate (Example):** To lock in an interest rate for 1926-1927 in 1925:
    1.  **Buy** a certain number of two-year discount bonds maturing in 1927.
    2.  **Short sell** a one-year discount bond maturing in 1926.
    This combination effectively creates a synthetic investment that yields the forward rate for the future period.
- **Key Takeaway:** The concept of forward rates, though seemingly simple, wasn't explicitly formalized or widely understood until someone (possibly Hicks) pointed out that these future rates could be determined and locked in using existing bond prices. This implies that even without a specific "forward rate" listed, the market implicitly provides it through the prices of bonds with different maturities.

## <a id='toc1_7_'></a>[Inflation & Interest Rates](#toc0_)

- **Nominal** rate $r_{\text{money}}$ quoted in dollars (but is *not* corrected for inflation)
- **Real** rate $r_{\text{real}}$ quoted in terms of the market baskets (that underlies the consumer price index/inflation rate)
- Nominal rate usually greater than real rate

$$
\begin{align}
    (r + r_{\text{money}}) &=         (1 + r_{\text{real}})(1 + i) \\ 
    r_{\text{money}}       &\tilde{=} r_{\text{real}} + i
\end{align}
$$

- $\tilde{=}$ means "isomorphic" $=$ "the same in whatever context we are"
- $i =$ rate of inflation
- $r_{\text{money}} =$ nominal rate
- $r_{\text{real}} =$ real rate

So just in simple terms, if you are investing money at 3% for next year and the inflation rate, consumer price index, is going up at 3% what is your real rate? How much are you making in real terms? $\rightarrow 0$ 

Thus, intuitively it would be
$$
    r_{\text{real}} = r_{\text{money}} - r_{\text{inflation}}
$$

## <a id='toc1_8_'></a>[Leverage](#toc0_)

<span style="color:brown">**$\textbf{Leveraging}$**</span> $=$ putting more money into the asset than you have
- if a company or an individual borrows money to buy assets

The start of the 2008-9 world financial crisis had to do with home buyers in U.S. and elsewhere borrowing to buy homes
- China today is a highly leveraged economy, arousing concerns
  - corporate debt, borrowing by corporations in China. It is 160% of GDP in China. 
  - In the US, it's only 70%. <span style="color:red">CHECK</span>
- Debt leads to bankruptcies, possible world crisis
  - if you have no debt, you normally don't go bankrupt
  - bankruptcy occurs when your creditors are after you for non-payment and normally, you would just pay them if you had the money

In a depression, wealth gets distributed towards the pessimits (creditors) so in a weighted sense they get more votes now. So you've now rewarded the pessimists and the world is being run by the pessimists.

Irving Fisher *"The Debt-Deflation Theory of Great Depressions"*, **Econometrica**, 1933
- Deflation redistributes real wealth from debtors (optimists/risk-takers) to creditors (pessimists/conservatives)
  - in a deflation, debtors get beaten down
  - The optimists have less wealth in real terms and the pessimists have more because now the $100 that they loaned is worth a lot more.
- Creditors tend to be the more cautious
- (Recent crisis has not been mostly deflationary, but inflation has falled short of expectations)

Leveraging increases risk.

- **Leveraging:** Borrowing money to invest in assets, increasing potential gains but also amplifying potential losses. For example, borrowing $100 to buy $200 in stocks means doubling profits if the stock doubles, but being wiped out if it falls by 50%.
- **Economic Risk:** Highly leveraged economies (like China, with corporate debt at 160% of GDP vs. 70% in the US) are more vulnerable to market downturns and corrections, as debt can lead to bankruptcy.
- **Irving Fisher's Debt-Deflation Model (1933):** Fisher argued that during deflation, the real value of debt increases, disproportionately harming optimistic debtors and benefiting pessimistic creditors. This shift in wealth to risk-averse individuals can exacerbate economic depressions.
- **Recent Crisis (2008/09):** While not experiencing severe deflation like 1933, the recent crisis still led to lower-than-expected inflation, effectively redistributing wealth from borrowers to lenders.
- **Lack of Inflation Indexation:** A puzzling aspect of human behavior is the general reluctance to index debt to inflation, meaning unexpected price changes can still cause wealth redistribution between debtors and creditors.
- **Leverage Cycle (John Geanakoplos):** Leverage varies over time; it became extremely high in the US housing market before the 2008 crisis, driven by an optimistic bias that home prices wouldn't fall.
- **Debt's Morality:** The speaker concludes that debt itself isn't immoral; it allows individuals and companies to smooth consumption and invest, even for essential needs or life events. However, it requires careful management.

Before the 2008-9 financial crises people borrowed up to 97% of the money of houses prices.

- **Dilution**: When new shares are issued, existing shareholders own a smaller % of the company.
- **Pecking Order Theory** (Stewart Myers):
  - Companies prefer to fund projects in this order:
    1. Retained earnings
    2. Debt
    3. New equity (least preferred — reduces control and profit share)
- **Myers (1984)**: Claimed most firms rarely issue new shares, supporting a “Marx was right” view — share trading is mostly secondary market activity.
- **Critique (Fama & French)**: Myers' time period was unusual (1973–82 stock market slump). Equity issuance rose again later — so share prices *do* influence corporate fundraising.
- **Bottom line**: Issuing shares affects control and ownership, so boards are cautious — but share prices still matter for raising capital when conditions are right.

Note that **all common shareholders are *equal***.

> most of the return people have gotten historically from stocks is in **dividends**, not in capital gains.

- what your really pricing in the stock market is a *claim on dividends*
- other than the PE-ratio there is also the PD-ratio (price-dividend):
$$
    \text{PD} = \frac{\text{DividendsPerShare}}{\text{PricePerShare}}
$$

If hypothetically a company pays out all his dividends then $PE = PD$.

<span style="color:brown">**$\textbf{Efficient Markets Idea}$**</span>: 
- $=$ <span style="color:blue">Price of a share is the present discounted value of its expected future dividends</span>

If companies have low or high PE ratios is only depending on 
1. $g =$ growth rate of earnings 
2. $r =$ interest/discount rate

Note
- If a company has **low PE** $\Rightarrow$ either 
  - $r$ is high, or
  - $g$ is low

<span style="color:brown">**$\textbf{Security Markets Line}$**</span>: 
- $=$ the expected return on a stock is a function of it's covariance with the market

Thus, that should mean: 
- <span style="color:blue">Companies whose return covaries with the market should tend to have **low PE ratios**</span>
  - $=$ they're riskier in the correct sense that CAPM provides
  - $=$ they're riskier in that they co-vary with the market

In general: 
- <span style="color:blue">High-growth companies should ten dot have high PE</span>
- <span style="color:blue">Low-growth companies should ten dot have low PE</span>

## <a id='toc2_9_'></a>[Why do Firms Pay Dividends?](#toc0_)

- Even when there was a stronger tax advatnage to capital gains, firms paid dividend
- **Dividend Signalling**
  - use dividends to proof something
  - by raising dividends, firm shows it can court bankruptcy
  - firms wanna show: we are they, we have real money
  - problem: alternative signalling methods are cheaper tax-wise

<span style="color:brown">**$\textbf{Lintner Model of Dividends}$**</span>: 
- Firms just gradually adjust dividends toward earnings per share
- So they have some target payout ratio
- $DIV_t - DIV_{t-1} = \rho(\tau \times EPS_t - DIV_{t-1})$
  - $\rho =$ adjustment rate, $\rho \in (0, 1)$
  - $\tau =$ target ratio, $\tau \in (0, 1)$

$$
    DIV_t = \rho\tau\sum_{k=1}^{\infty}(1 - \rho)^k EPS_{t-k}
$$
